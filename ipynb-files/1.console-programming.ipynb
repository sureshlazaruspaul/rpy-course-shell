## Advanced computation using math package

### 1. item computations

```python
# import packages and dependencies
import math

input_value = 10
```

```python
# log computation

print(
    f"Log value of {input_value} is {math.log(input_value)}"
)
```

```python
# square computation

print(
    f"Squared value of {input_value} is {math.pow(input_value, 2)}"
)
```

```python
# square-root computation

print(
    f"Sq-root value of {input_value} is {math.sqrt(input_value)}"
)
```

## Advanced computation using numpy package

```python
# import packages and dependencies
import numpy as np

input_value = 10
```

```python
# simple computation

print(
    f"Log value of {input_value} is {np.log(input_value)}"
)
```

```python
# vector computations

# input a vector
set_of_numbers = [1, 2, 4, 7, 11, 19, 29, 34, 81]
```

```python
np.sum(set_of_numbers)
```

```python
np.prod(set_of_numbers)
```

```python
np.cumsum(set_of_numbers)
```

```python
np.cumprod(set_of_numbers)
```
