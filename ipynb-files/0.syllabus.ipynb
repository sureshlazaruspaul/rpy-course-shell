COURSE DESCRIPTION

This course introduces students to various tools in business analytics development and reporting. Students are introduced to the basics of programming with R and Python to develop a working knowledge of how to acquire, manage, and treat data.

***

LEARNING OUTCOMES

- Understand and apply the main concepts used in data science and business analytics using functions and tools developed in Python
- Understand and use R functions in business data analysis
- Understand and apply basic coding and programing with tools such as R and Python

## AI log + AI Reflection statement

### USE OF AI FOR CODING

The *ethical* use of AI as a coding companion is encouraged. Its usage should align with the course's learning standards and should not serve as a substitute for the coding skills taught in class. Put simply, if you do not understand the code that AI recommends, DO NOT USE IT. There are explicit points for presenting your homework during class, and if you are unable to explain your own code, you will receive a ZERO for that assignment.


-   Proper logging is mandatory and will help in writing the AI Reflection statement at the end of term.
    -   Log every AI interaction during the assignment.
    -   *What were you trying to accomplish?* (e.g., "Debug syntax error," "Understand how loops works," "Generate boilerplate code for a class")
    -   *Which tool did you use?* (e.g., ChatGPT, Claude, Gemini, GitHub Copilot, other)
    -   *Type of Assistance*
        -   [ ] Explaining a concept
        -   [ ] Debugging/fixing errors
        -   [X] Code generation (wrote new code)
        -   [ ] Code review/optimization
        -   [ ] Research/documentation lookup
        -   [ ] Other: _______________
    -   *What did you learn from this interaction?* Did it help you understand the concept better, or just solve the immediate problem, or did it confuse you more?
    -   *Approximate time on this AI interaction*: _____ minutes
    -   AI work is highly detectable in homework. Using AI and ***not*** citing it will earn a ZERO for that homework. This detection will be compared to your Statement of AI Use to determine any grading penalty. This strictness reflects the university's policy on academic integrity: it is impossible to assign a grade for homework that does not reflect your own contribution but rather that of an AI, especially if you do not log it.
    -   End of term, you are expected to turn in a detailed reflection on overall AI use patterns during the course. You may draw insights from your individual homework logs to write this report.
